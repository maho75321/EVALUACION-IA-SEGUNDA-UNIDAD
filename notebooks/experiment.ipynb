## 1. Configuración del experimento

## 2. Carga y exploración del dataset

Se cargan CIFAR-10, se visualizan 16 imágenes y se crean
las particiones de entrenamiento, validación y prueba.


```python
(
    cargador_entrenamiento,
    cargador_validacion,
    cargador_prueba,
    clases,
) = cargar_datos(mostrar_imagenes=True)

print("Clases:", clases)
print("Tamaño de entrenamiento:",
      len(cargador_entrenamiento.dataset))
print("Tamaño de validación:",
      len(cargador_validacion.dataset))
print("Tamaño de prueba:",
      len(cargador_prueba.dataset))
```

## 3. Arquitectura y tensores 4D

Se verifican las dimensiones de entrada, las capas convolucionales,
el aplanado y la salida de diez clases.


```python
comprobar_dimensiones()

modelo = CNNClasificador()
imagenes, etiquetas = next(iter(cargador_entrenamiento))

with torch.no_grad():
    predicciones = modelo(imagenes)

print("Forma de las imágenes:", tuple(imagenes.shape))
print("Forma de las etiquetas:", tuple(etiquetas.shape))
print("Forma de las predicciones:", tuple(predicciones.shape))

assert predicciones.shape == (imagenes.shape[0], 10)
```
